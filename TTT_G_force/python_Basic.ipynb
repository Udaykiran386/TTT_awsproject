```python
# Writing using 'with' block
with open("example.txt", "w", encoding="utf-8") as f:
    f.write("Line 1\nLine 2\n")
```

```python
# Reading using 'with' block
with open("example.txt", "r", encoding="utf-8") as f:
    content = f.read()
    print(content)
```

```python
# 2. Read fixed chunk (e.g., 10 characters)
with open("example.txt", "r", encoding="utf-8") as f:
    chunk = f.read(10)
    print(chunk)
```

```python
# 3. Read line-by-line using a loop (Memory Efficient for large files)
with open("example.txt", "r", encoding="utf-8") as f:
    for line in f:
        print(line.strip())
```

```python
# 4. Read single line
with open("example.txt", "r", encoding="utf-8") as f:
    first_line = f.readline()
    print(first_line)
```

```python
# 5. Read all lines into a list
with open("example.txt", "r", encoding="utf-8") as f:
    lines = f.readlines()
    print(lines)
```

```python
# 1. Overwrite mode ('w')
with open("example.txt", "w", encoding="utf-8") as f:
    f.write("This replaces existing content.\n")
```

```python
# 2. Append mode ('a')
with open("example.txt", "a", encoding="utf-8") as f:
    f.write("This adds to the end of the file.\n")
```

```python
# 3. Write multiple lines from a list
lines = ["First line\n", "Second line\n", "Third line\n"]
with open("example.txt", "w", encoding="utf-8") as f:
    f.writelines(lines)
```

```python
# Reading binary data
with open("daylight-environment-forest-459225.jpg", "rb") as f:
    binary_data = f.read()
    print(binary_data)
```

```python
with open("example.txt", "w+", encoding="utf-8") as f:
    f.write("Hello World")
    
    print(f.tell())  # Output: 11 (end of file)
    
    # Move cursor back to beginning
    f.seek(0)
    print(f.tell())  # Output: 0
    
    # Read from start
    print(f.read())  # Output: Hello World
```

```python
from pathlib import Path

path = Path("example.txt")

# Quick write
path.write_text("Hello using pathlib!", encoding="utf-8")

# Quick read
content = path.read_text(encoding="utf-8")

# Using 'open()' via pathlib object
with path.open("r", encoding="utf-8") as f:
    data = f.read()
```
