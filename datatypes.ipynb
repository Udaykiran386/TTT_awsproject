Identify the data type of the following literals:

a = 3.0

b = "100"

c = (5,)

d = {5}

e = True + 2

```python
a = 3.0

b = "100"

c = (5,)

d = {5}

e = True + 2

print(type(a))
print(type(b))
print(type(c))
print(type(d))
print(type(e))
```

upper() / lower(): Converts string casing.

```python
text = "mY Name iS Raga"
print(text.upper())  
```

strip(): Removes leading and trailing whitespace.

```python
print(' data'.strip())
print(' data '.strip())
print('data '.strip())
print('   data'.strip())
```

replace(old, new): Replaces substrings with a new value.

```python
phrase = "I like Java"
print(phrase.replace("Java", "Python"))
```

split(separator): Splits a string into a list of substrings.

```python
print("apple,banana,cherry".split(","))
print("apple banana cherry".split(" "))
print("apple_banana_cherry".split("_"))
print("apple,banana,cherry".split("a"))
```

join(iterable): Joins elements of an iterable into a single string.

```python
words = ["Python", "is", "fun"]
print(" ".join(words))
```
