```python
# prime numbers

def is_prime(n):
    if n < 2:
        return False
    return all(n % i != 0 for i in range(2, int(n ** 0.5) + 1))

new_list = [1, 3, 21, 17, 44, 65]
prime_list = list(filter(lambda x: is_prime(x), new_list))

print(prime_list)
```

```python
nums = [1,2,4,6,7]
missing = [x for x in range(min(nums), max(nums)+1) if x not in nums]
print(missing)
```

```python
nums = range(100, 1000)
armstrong = [n for n in nums if n == sum(int(d)**3 for d in str(n))]
print(armstrong)
```

```python
nums = [121, 123, 454, 789]
palindromes = list(filter(lambda n: str(n) == str(n)[::-1], nums))
print(palindromes)
```

```python
class Student:
    school_name = "Delhi Public School"   # class variable - same for everyone
    total_students = 0                     # class variable used as a counter

    def __init__(self, name, roll_no):
        self.name = name           # instance variable - unique per student
        self.roll_no = roll_no
        Student.total_students += 1


s1 = Student("Meera", 1)
s2 = Student("Arjun", 2)
print(s1.school_name)
print(s2.school_name)

print(s1.school_name, s2.school_name)   # Delhi Public School / Delhi Public School

s1.school_name = "St. Xavier's"    # change once, at the class level
print(s1.school_name, s2.school_name)   # both update automatically -> St. Xavier's / St. Xavier's

print(Student.total_students)           # 2 - a class variable tracking all instances ever created
```
