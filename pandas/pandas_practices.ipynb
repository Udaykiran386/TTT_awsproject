# Step 2: Data Selection & Filtering

# Step 3: Data Cleaning

```python
import numpy as np
# 1. Handle missing values
# (Simulating missing values in 'pop')
df_dirty = df.copy()
df_dirty.loc[0:2, "pop"] = np.nan
print(df_dirty)
```

```python
# Drop rows with missing values
df_dropped = df_dirty.dropna()
print(df_dropped)
```

```python
# Fill missing values with 0
df_filled = df_dirty.fillna(value=0)
print(df_filled.head(4)[["country", "year", "pop"]])
```

```python
# 2. Rename columns
df_renamed = df.rename(columns={
    "pop": "population", 
    "gdpPercap": "gdp_per_capita"
})
print("Renamed columns:", df_renamed.columns.tolist())
# print(df_renamed)
# print(df)             # check this both are same then u can find the difference
```

```python
# 3. Change data types
# Convert float population back to integer
df["pop"] = df["pop"].astype("int64")
print("Updated Dtype:", df["pop"].dtype)
```

# Step 4: Transformations

```python
# 1. Apply functions (Convert life expectancy years to days)
df["lifeExp_days"] = df["lifeExp"].apply(lambda x: x * 365.25)

# 2. Vectorized operation & new column creation (Calculate total GDP)
df["gdp_total"] = df["gdpPercap"] * df["pop"]

# 3. Vectorized scaling (Convert total GDP to millions)
df["gdp_total_millions"] = df["gdp_total"] / 1_000_000

print(df[["country", "year", "lifeExp", "lifeExp_days", "gdp_total_millions"]].head(3))
```

# Step 5: Grouping & Aggregation

```python
# 1. Group by continent and compute mean life expectancy
continent_life_exp = df.groupby("continent")["lifeExp"].mean()
print(continent_life_exp)
```

```python
# 2. Advanced multi-column aggregation
aggregated_summary = df.groupby("continent").agg({
    "lifeExp": ["mean", "max"],
    "gdpPercap": ["mean", "max"]
})
print(aggregated_summary)
```

# Step 6: Joins & Merges

```python
# Create sample datasets for merging
df_continents = df[["country", "continent"]].drop_duplicates().head(5)
df_2007_stats = df[(df["year"] == 2007) & (df["country"].isin(df_continents["country"]))][["country", "lifeExp", "gdpPercap"]]
print(df_continents)
print(df_2007_stats)
```

```python
# 1. Merge datasets on common key 'country'
merged_df = pd.merge(df_continents, df_2007_stats, on="country", how="inner")
print(merged_df)
```

```python
# 2. Concatenate datasets vertically
df_1952_afg = df[(df["year"] == 1952) & (df["country"] == "Afghanistan")][["country", "year", "lifeExp"]]
df_2007_afg = df[(df["year"] == 2007) & (df["country"] == "Afghanistan")][["country", "year", "lifeExp"]]

concatenated_df = pd.concat([df_1952_afg, df_2007_afg], ignore_index=True)
print(concatenated_df)
```

# Step 7: Sorting & Ranking

```python
# Filter data for year 2007 for analysis
df_2007 = df[df["year"] == 2007].copy()

# 1. Sort values by GDP per capita (Descending)
df_sorted = df_2007.sort_values(by="gdpPercap", ascending=False)
print(df_sorted[["country", "continent", "gdpPercap"]].head(3))
```

```python
# 2. Rank countries by Life Expectancy in 2007 (Rank 1 = Highest Life Expectancy)
df_2007["lifeExp_rank"] = df_2007["lifeExp"].rank(ascending=False, method="min")
print(df_2007.sort_values("lifeExp_rank")[["country", "lifeExp", "lifeExp_rank"]].head(3))
```

# Step 8: Advanced Operations

```python
# 1. Pivot table: Mean life expectancy by continent (rows) and selected years (columns)
pivot_df = df[df["year"].isin([1952, 1977, 2007])].pivot_table(
    values="lifeExp", 
    index="continent", 
    columns="year", 
    aggfunc="mean"
)
print(pivot_df)
```

```python
# 2. Window function: 3-period rolling average of life expectancy for Afghanistan
df_afg = df[df["country"] == "Afghanistan"].copy()
df_afg["lifeExp_rolling_3"] = df_afg["lifeExp"].rolling(window=3).mean()
print(df_afg[["country", "year", "lifeExp", "lifeExp_rolling_3"]].head(6))
```
