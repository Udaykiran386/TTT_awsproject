What Python is

Python is a high-level programming language, which means it lets you write instructions in a human-readable form instead of low-level machine instructions. It is designed to be simple, readable, and productive. Python is widely used for web development, automation, data science, AI, cybersecurity, and scripting.

Why it is high-level

Its syntax is readable and looks closer to plain English than to hardware instructions.

It hides low-level details like memory access and processor operations, so you can focus on logic instead of machine internals.

It is portable, meaning the same Python code can usually run on different operating systems with little or no change.

It supports high-level built-in data structures and dynamic typing, which reduce the amount of code you need to write.

Why Python is called interpreted

Python is called an interpreted language because your code is not directly turned into machine code in one separate compilation step before execution. Instead, a Python interpreter reads your code, checks it, converts it into an intermediate form called bytecode, and then executes it through the Python virtual machine. The interpreter acts like a translator between your code and the computer’s machine language.

A simple way to think about it:

You write Python code.

The interpreter reads it.

It converts it into bytecode.

The bytecode is executed by Python’s runtime system.

How Python runs your code

When you run a Python file, several things happen internally:

Lexical analysis: Python breaks the code into small pieces called tokens.

Parsing: It checks whether the tokens follow Python grammar rules.

AST creation: It builds an Abstract Syntax Tree, which represents the structure of your program.

Bytecode compilation: The AST is converted into bytecode.

Execution: The bytecode is run by the Python runtime.

So although Python feels “line by line” to beginners, internally it is a little more structured than that.

Interactive mode vs script mode

Python can be used in two main ways:

Interactive mode: You type commands one by one and see results immediately. This is useful for learning, quick testing, and debugging.

Script mode: You write code in a .py file and run the whole program. This is how real applications are usually built.

Python syntax basics

Python syntax is the set of rules for writing valid Python code. Some important basics:

Indentation matters: Python uses spaces to define blocks of code.

Variables are dynamically typed: You do not need to declare a type before using a variable.

Statements and expressions: A statement performs an action, while an expression produces a value.

Comments: Used to explain code and are ignored by Python.

Shallow copy

A shallow copy duplicates only the top level of a compound object. If the object contains nested lists, dictionaries, or other mutable objects, those inner objects are still shared between the original and the copy.

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)

b[0][0] = 99
print(a)  # [[99, 2], [3, 4]]
print(b)  # [[99, 2], [3, 4]]
```

Deep copy

A deep copy recursively copies the entire object structure, including all nested objects. After that, the original and the copy are independent, so changes in one do not affect the other.

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)

b[0][0] = 99
print(a)  # [[1, 2], [3, 4]]
print(b)  # [[99, 2], [3, 4]]
```

In Python, implicit conversion happens automatically, while explicit conversion is done by the programmer using functions like int(), float(), str(), etc. Python does implicit conversion when it can safely convert one type to another, and explicit conversion when you tell it exactly what to do.

Implicit conversion

Python converts types on its own during expressions.

```python
a = 10
b = 2.5
c = a + b
print(c)      # 12.5
print(type(c)) # float
```

Explicit conversion

You manually convert one type into another.

```python
x = "10"
y = int(x)
print(y)      # 10
print(type(y)) # int
```

List functions and methods

Common list methods include append(), extend(), insert(), remove(), pop(), clear(), index(), count(), sort(), reverse(), and copy().

```python
nums = [1, 2, 3]
nums.append(4)       # [1, 2, 3, 4]
nums.insert(1, 10)   # [1, 10, 2, 3, 4]
nums.remove(10)      # [1, 2, 3, 4]
nums.pop()           # 4
nums.sort()          # [1, 2, 3]
```

Tuple functions and methods

Tuples mainly support count() and index() because they cannot be modified after creation.

```python
t = (1, 2, 1, 3)
print(t.count(2))    # 2
print(t.index(3))    # 3
```

Dictionary functions and methods

Common dictionary methods include get(), keys(), values(), items(), update(), pop(), popitem(), clear(), copy(), and setdefault().

len() works on dictionaries too, and it returns the number of key-value pairs.

```python
student = {"name": "Uday", "age": 22}

print(student.get("name"))     # Uday
print(student.keys())           # dict_keys(['name', 'age'])
print(student.values())         # dict_values(['Uday', 22])
print(student.items())          # dict_items([('name', 'Uday'), ('age', 22)])

student.update({"city": "Hyderabad"})
print(student)                  # {'name': 'Uday', 'age': 22, 'city': 'Hyderabad'}
```

Quick difference

List: ordered, mutable, supports many update methods.

Tuple: ordered, immutable, very few methods.

Dictionary: key-value mapping, mutable, has key-based methods.

Most useful ones to remember

List: append(), insert(), remove(), pop(), sort(), reverse().

Tuple: count(), index().

Dictionary: get(), keys(), values(), items(), update(), pop().

File Handling

File handling in Python comes down to one core idea: opening a file, performing an action (read, write, or append), and closing it.

Python makes this straightforward using the built-in open() function. Let me walk you through everything from basic text files to JSON and CSV, with clear examples for each step.

```python
with open("example.txt","w") as file:
    file.write("Hello, World!\nWelcome to python File Handling.")
```

```python
with open("example.txt","r") as file:
    content=file.read()
    print("---Full Concept---")
    print(content)
```

```python
with open("example.txt", "r") as file:
    print("--- First Line Only ---")
    print(file.readline().strip())
```

```python
with open("example.txt", "r") as file:
    print("--- Iterating Line by Line ---")
    for line in file:
        print(line.strip())
```

```python
with open("example.txt", "w") as file:
    file.write("This line overwrote everything previously in example.txt.\n")
```

```python
with open("example.txt", "a") as file:
    file.write("This line is appended at the bottom!\n")
    file.writelines([
        "Line A\n",
        "Line B\n"
    ])
```

```python
import json

data = {
    "name": "Alex",
    "skills": ["Python", "Data Science"],
    "is_active": True
}

# Writing JSON
with open("user.json", "w") as file:
    json.dump(data, file, indent=4)
```

```python
# Reading JSON
with open("user.json", "r") as file:
    loaded_data = json.load(file)
    print("Loaded JSON:", loaded_data)
```

```python
import csv

# Writing CSV
with open("data.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["ID", "Name", "Role"])
    writer.writerow([1, "Sarah", "Developer"])
    writer.writerow([2, "James", "Designer"])
```

```python
with open("data.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```
