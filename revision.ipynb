2. List & Dictionary Comprehensions
   
Comprehensions allow you to transform lists or dictionaries in a single, clean line of code. In PySpark, you'll use these constantly to clean column names, filter list attributes, or generate programmatic SQL expressions.

```python
# Example: Standard loop vs. List Comprehension
# Imagine we have a list of numbers and want to multiply each number by 2:
# --- TRADITIONAL WAY (FOR LOOP) ---
numbers = [1, 2, 3, 4, 5]
doubled_numbers = [] # Step 1: Create empty list

for num in numbers: # Step 2: Loop through each item
    result = num * 2
    doubled_numbers.append(result) # Step 3: Add to list

print(doubled_numbers) # Output: [2, 4, 6, 8, 10]


# --- PYSPARK / PYTHONIC WAY (LIST COMPREHENSION) ---
# Syntax: [expression for item in list]

print(list(num * 2 for num in numbers)) # Output: [2, 4, 6, 8, 10]
```

```python
# Adding a Filter (if condition)
# Suppose we only want to keep numbers greater than 2:
numbers = [1, 2, 3, 4, 5]

# Syntax: [expression for item in list if condition]
filtered_numbers = [num * 2 for num in numbers if num > 2]

print(filtered_numbers) # Output: [6, 8, 10]
```

Lesson 3: Lambda Functions & Functional Built-ins

Normally, you write a function using def. A lambda function is a short, 1-line anonymous function with no name.

```python
# Standard def vs lambda

# Standard function
def add_ten(x):
    return x + 10

# Equivalent Lambda function
# Syntax: lambda inputs : output_expression
add_ten_lambda = lambda x,y: x + 10 + y

print(add_ten(5))        # Output: 15
print(add_ten_lambda(5,10)) # Output: 15
```

Functional Built-in 1: map()
The built-in map(function, iterable) takes a function and applies it to every single item in a list.  

```python
salaries = [50000, 60000, 75000]

# Give everyone a $5,000 bonus using map() and lambda
# Note: list() converts map's internal generator output back into a readable list
updated_salaries = list(map(lambda salary: salary + 5000, salaries))

print(updated_salaries) # Output: [55000, 65000, 80000]
```

Functional Built-in 2: filter()
The built-in filter(function, iterable) keeps items where the function evaluates to True and discards the rest.

```python
ages = [15, 22, 18, 40, 12, 31]

# Keep only adults (age >= 18)
adults = list(filter(lambda age: age >= 18, ages))

print(adults) # Output: [22, 18, 40, 31]
```

Lesson 4: Try-Except Error Handling

Data pipelines break when they encounter corrupt data (like trying to divide by zero or parsing text as a number). try-except prevents your pipeline from crashing.

```python
raw_data_inputs = ["100", "200", "corrupted_text", "400"]
cleaned_data = []

for item in raw_data_inputs:
    try:
        # Try converting string to integer
        number = int(item)
        cleaned_data.append(number)
    except ValueError:
        # If conversion fails (e.g. for "corrupted_text"), execution jumps here instead of crashing!
        print(f"Warning: Skipping invalid item -> '{item}'")

print("Final Cleaned Data:", cleaned_data)
# Output:
# Warning: Skipping invalid item -> 'corrupted_text'
# Final Cleaned Data: [100, 200, 400]
```

Loop through raw_employees using a try-except block to ignore "INVALID_SALARY".

Use filter() with a lambda function to keep only employees earning over $50,000.

```python
raw_employees = [
    ("Alice", 80000),
    ("Bob", "INVALID_SALARY"),
    ("Charlie", 45000),
    ("David", 120000)
]

new_employees = []
for name, salary in raw_employees:
    try:
        salary = int(salary)  # Attempt to convert salary to integer
        new_employees.append((name, salary))
    except ValueError:
        print(f"Warning: Skipping new employee with invalid salary -> '{name}: {salary}'")

print("Final Cleaned Employee Data:", list(filter(lambda x: x[1] > 50000, new_employees)))
```

```python
raw_sensor_readings = [
    ("Sensor_A", "98.6", "22.4"),
    ("Sensor_B", "NULL", "19.8"),
    ("Sensor_C", "101.2", "ERR"),
    ("Sensor_D", "97.4", "21.0")
]
cleaned_readings = []
for sensor, temp,humidity in raw_sensor_readings:
    try:
        temp=float(temp)
        humidity=float(humidity)
        cleaned_readings.append((sensor, temp, humidity))
    except ValueError:
        print(f"Warning: Skipping sensor reading for {sensor} due to invalid data -> Temp: '{temp}', Humidity: '{humidity}'")

print("Final Cleaned Sensor Readings:", cleaned_readings)
```
